```python
# Training data
data = [
    ["A", 2, 40, 45],
    ["B", 3, 45, 50],
    ["C", 4, 60, 62],
    ["D", 5, 65, 68],
    ["E", 7, 80, 82]
]

# Test observation
test_study = 4.5
test_assignment = 63

# K value
k = 3

print("Training Data:")
print("Student  Study Hours  Assignment  Final Marks")

for row in data:
    print(row[0], "       ", row[1], "          ", row[2], "          ", row[3])

print("\nTest Observation:")
print("(Study Hours, Assignment Marks) = (4.5, 63)")

print("\nStep 1: Calculate Euclidean Distance")
print("Formula:")
print("d = sqrt((x1-x2)^2 + (y1-y2)^2)")

# Calculate distances
distances = []

for row in data:
    student = row[0]
    study = row[1]
    assignment = row[2]
    final = row[3]

    distance = ((study - test_study) ** 2 +
                (assignment - test_assignment) ** 2) ** 0.5

    distances.append([distance, student, final])

    print("\nStudent", student)
    print("d = sqrt((", study, "- 4.5)^2 + (", assignment, "- 63)^2)")
    print("d =", round(distance, 2))

# Sort distances
distances.sort()

print("\nStep 2: Arrange distances from smallest to largest")

for item in distances:
    print(item[1], "-> Distance =", round(item[0], 2),
          ", Final Marks =", item[2])

# Select K nearest neighbors
print("\nStep 3: Select K =", k, "nearest neighbors")

total = 0

for i in range(k):
    print(distances[i][1], "->", distances[i][2])
    total = total + distances[i][2]

# Calculate average
print("\nStep 4: Calculate Average")

print("Prediction = (", distances[0][2], "+",
      distances[1][2], "+", distances[2][2], ") /", k)

prediction = total / k

print("Prediction =", round(prediction, 2))

print("\nFinal Predicted Final Marks =", round(prediction, 2))
```
