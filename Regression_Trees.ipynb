```python
# Step 1: Training data
X = [1, 2, 3, 4, 5, 6, 7, 8]
Y = [10, 12, 15, 18, 20, 25, 28, 30]


# Calculate Mean Squared Error
def mse(values):
    mean = sum(values) / len(values)

    error = 0
    for value in values:
        error = error + (value - mean) ** 2

    return error / len(values)


# Find the best split
def find_best_split(x, y):

    best_split = None
    best_error = float('inf')

    for split in x:

        left_y = []
        right_y = []

        for i in range(len(x)):
            if x[i] <= split:
                left_y.append(y[i])
            else:
                right_y.append(y[i])

        # Ignore invalid splits
        if len(left_y) == 0 or len(right_y) == 0:
            continue

        # Error after split
        error = (len(left_y) * mse(left_y) +
                 len(right_y) * mse(right_y)) / len(y)

        if error < best_error:
            best_error = error
            best_split = split

    return best_split, best_error


# Build the regression tree
def build_tree(x, y, depth):

    # Step 6: Stop rule
    # Stop if only 2 observations are left
    if len(y) <= 2:
        mean = sum(y) / len(y)
        return mean

    # Step 2, 3 and 4: Find best split
    split, error = find_best_split(x, y)

    # If no split is possible
    if split is None:
        mean = sum(y) / len(y)
        return mean

    print("Depth:", depth, " Split:", split, " Error:", error)

    # Step 5: Create child nodes
    left_x = []
    left_y = []

    right_x = []
    right_y = []

    for i in range(len(x)):
        if x[i] <= split:
            left_x.append(x[i])
            left_y.append(y[i])
        else:
            right_x.append(x[i])
            right_y.append(y[i])

    # Step 5: Repeat on child nodes
    left = build_tree(left_x, left_y, depth + 1)
    right = build_tree(right_x, right_y, depth + 1)

    return (split, left, right)


# Build the tree
tree = build_tree(X, Y, 0)

print("\nFinal Tree:")
print(tree)


# Step 8: Prediction using mean value in leaf
def predict(tree, value):

    # If tree is a leaf, return its mean
    if not isinstance(tree, tuple):
        return tree

    split, left, right = tree

    if value <= split:
        return predict(left, value)
    else:
        return predict(right, value)


# Test prediction
new_value = 6

prediction = predict(tree, new_value)

print("\nPrediction for", new_value, "=", prediction)
```
